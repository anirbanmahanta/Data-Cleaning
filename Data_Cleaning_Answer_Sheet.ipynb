```python
import pandas as pd
import numpy as np
df = pd.read_csv('/content/healthcare_data_cleaning_dataset.csv')
df.head()
```

#**Q1. Missing Data Identification**

Scenario:
The hospital suspects incomplete patient records.

Task:
* Identify missing values in each column
* Calculate percentage of missing data



```python
missing_values = df.isnull().sum()
percentage_missing = (missing_values / len(df)) *100
print("Missing Values in Each Column:")
print(missing_values)
print("\nPercentage of Missing Data in Each Column:")
print(percentage_missing)
```

#**Q2. Handling Missing Age**

Scenario:
 Age is critical for medical analysis, but some values are missing.

Task:
* Replace missing Age values with an appropriate method
* Justify your choice (mean/median)

```python
# Check missing values in Age
print("Missing Age values before:")
print(df['Age'].isnull().sum())

# Calculate median age
median_age = df['Age'].median()

# Replace missing Age values with median
df['Age'].fillna(median_age, inplace=True)

# Verify missing values are removed
print("\nMissing Age values after:")
print(df['Age'].isnull().sum())

print("\nMedian Age used for imputation:", median_age)
```

Age is a numerical variable and often contains outliers (very young or very old patients). Therefore, Median Imputation is usually preferred over Mean Imputation because the median is less affected by extreme values and provides a more robust estimate of the central tendency.

# **Q3. Handling Missing Treatment Cost**

Scenario:
 Treatment cost is highly skewed due to expensive treatments.

Task:
* Handle missing Treatment_Cost values
* Choose the correct imputation method and explain why

```python
# Check missing values
print("Missing Treatment_Cost values before imputation:")
print(df['Treatment_Cost'].isnull().sum())

# Calculate median treatment cost
median_cost = df['Treatment_Cost'].median()

# Replace missing values with median
df['Treatment_Cost'] = df['Treatment_Cost'].fillna(median_cost)

# Verify missing values are removed
print("\nMissing Treatment_Cost values after imputation:")
print(df['Treatment_Cost'].isnull().sum())

print("\nMedian Treatment Cost used for imputation:", median_cost)
```

Missing values in the Treatment_Cost column were replaced using the median value of the column. Median imputation was selected because treatment costs are typically highly skewed, with a few patients having exceptionally high medical expenses. These extreme values can significantly influence the mean, making it a less reliable measure of central tendency. The median is robust to outliers and better represents a typical treatment cost, resulting in a more accurate and unbiased imputation strategy.

# **Q4. Duplicate Patient Records**

Scenario:
 Some patient records were entered multiple times.

Task:
* Identify duplicate rows
* Remove duplicates
* Compare dataset size before and after

```python
# Dataset Size Before
rows = df.shape[0]
columns = df.shape[1]
print("Dataset Size Before:")
print("Rows:", rows)
print("Columns:", columns)

# Duplicate Rows
duplicate_rows = df[df.duplicated()]
print("Duplicate Rows:")
print(duplicate_rows)

# Remove Duplicates
df.drop_duplicates(inplace=True)

# Dataset After Duplicates Remove
rows_after = df.shape[0]
columns_after = df.shape[1]
print("\nDataset Size After Duplicates Remove:")
print("Rows:", rows_after)
print("Columns:", columns_after)

# Number of Rows Removes
rows_removed = rows - rows_after
print("\nNumber of Rows Removed:", rows_removed)
```

# **Q5. Invalid Age Values (Data Quality Check)**

Scenario:
 Some patients have unrealistic age values (e.g., >100 or <0).

Task:
* Detect such records
* Decide whether to remove or correct them

```python
# Dataset before cleaning
print("Dataset before cleaning:" , df.shape)

# Detect invalid age values
invalid_age_rows = df[(df['Age'] > 100) | (df['Age'] < 0)]
print("\nInvalid Age Values:" , len(invalid_age_rows))

print("\nInvalid Age Values:")
print(invalid_age_rows)

# Remove Invalid Age
df = df[(df['Age'] <= 100) & (df['Age'] >= 0)]

# Dataset after cleaning
print("\nDataset after cleaning:" , df.shape)

# Number Of Record Remove
print("\nNumber Of Record Remove:" , len(invalid_age_rows))
```

# **Q6. Outlier Detection (Treatment Cost)**

Scenario:
 Extreme treatment costs are affecting analysis.

Task:
* Detect outliers using IQR method
* Display number of outliers

```python
# Q1 and Q3
q1 = df['Treatment_Cost'].quantile(0.25)
q3 = df['Treatment_Cost'].quantile(0.75)

# Calculate IQR
iqr = q3 - q1

# Bounds
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Number of Outlier
outliers = df[(df['Treatment_Cost'] < lower_bound) | (df['Treatment_Cost'] > upper_bound)]
print("\nNumber of Outliers:", len(outliers))

# Percentage of Outlier
percentage_outliers = (len(outliers) / len(df)) * 100
print("Percentage of Outliers:", percentage_outliers)
```

# **Q7. Outlier Treatment**

Scenario:
The business team wants to retain all records.

Task:
* Apply capping (Winsorization) on Treatment_Cost
* Use 5th and 95th percentile

```python
# 5th and 95th percentiles
lower_cap = df['Treatment_Cost'].quantile(0.05)
upper_cap = df['Treatment_Cost'].quantile(0.95)

print("5th Percentile:", lower_cap)
print("95th Percentile:", upper_cap)

# Create a copy for comparison
df_winsorized = df.copy()

# Apply Winsorization
df_winsorized['Treatment_Cost'] = df_winsorized['Treatment_Cost'].clip(lower=lower_cap, upper=upper_cap)

# Check results
print("\nOriginal Treatment Cost Summary:")
print(df['Treatment_Cost'].describe())

print("\nTreatment Cost Summary After Winsorization:")
print(df_winsorized['Treatment_Cost'].describe())
```

# **Q8. Transformation**

Scenario:
 Treatment cost is highly skewed.

Task:
* Apply log transformation
* Create a new column
* Compare before vs after distribution

```python
# Create a new column with log transformation
# log1p(x) = log(1 + x), handles zero values safely
df['Treatment_Cost_Log'] = np.log1p(df['Treatment_Cost'])

# Compare summary statistics
print("Original Treatment Cost Distribution")
print(df['Treatment_Cost'].describe())

print("\nLog-Transformed Treatment Cost Distribution")
print(df['Treatment_Cost_Log'].describe())

# Compare skewness
print("\nSkewness Before Transformation:")
print(df['Treatment_Cost'].skew())

print("\nSkewness After Transformation:")
print(df['Treatment_Cost_Log'].skew())

# Display first few rows
print("\nSample Data:")
print(df[['Treatment_Cost', 'Treatment_Cost_Log']].head())
```

# **Q9. Time-Based Missing Handling**

Scenario:
 Admission dates should follow a logical sequence.

Task:
* Sort data by Admission_Date
* Apply forward fill or backward fill where appropriate
* Justify your choice

```python
df['Admission_Date'] = pd.to_datetime(df['Admission_Date'])

# Sort by date
df = df.sort_values('Admission_Date')

# Check missing values before filling
print("Missing values before filling:")
print(df.isnull().sum())

# Forward Fill
df_ffill = df.ffill()

# Backward Fill
df_bfill = df.bfill()

print("Forward Fill Missing Values:")
print(df_ffill.isnull().sum())

print("\nBackward Fill Missing Values:")
print(df_bfill.isnull().sum())
```

The dataset was first sorted by Admission_Date to maintain the correct chronological order. Missing values were then handled using the Forward Fill (ffill) method. Forward fill propagates the most recent valid observation to subsequent missing entries, making it suitable for time-ordered data where current records are likely to resemble recent past records. This approach preserves temporal consistency and avoids using future information to fill past records. Backward fill (bfill) could also be used in situations where future observations are more appropriate, but forward fill was selected as the primary method because it aligns better with real-world sequential data behavior.
